# EDA with SQL - Completed
All ten SQL tasks from the capstone SQL lab are included below.


```python
import pandas as pd, sqlite3
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv")
con = sqlite3.connect(":memory:")
df.to_sql("SPACEXTABLE", con, index=False, if_exists="replace")
def q(sql):
    return pd.read_sql_query(sql, con)
```

## Task 1 - Unique launch sites


```python
q("SELECT DISTINCT Launch_Site FROM SPACEXTABLE")
```

## Task 2 - Five records where launch site begins with 'CCA'


```python
q("SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5")
```

## Task 3 - Total payload mass carried for NASA (CRS)


```python
q("SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload FROM SPACEXTABLE WHERE Customer = 'NASA (CRS)'")
```

## Task 4 - Average payload mass for F9 v1.1


```python
q("SELECT AVG(PAYLOAD_MASS__KG_) AS Avg_Payload FROM SPACEXTABLE WHERE Booster_Version = 'F9 v1.1'")
```

## Task 5 - First successful ground-pad landing


```python
q("SELECT MIN(Date) AS First_Success_Ground FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (ground pad)'")
```

## Task 6 - Successful drone-ship boosters with payload between 4000 and 6000 kg


```python
q("SELECT DISTINCT Booster_Version FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000")
```

## Task 7 - Mission outcome counts


```python
q("SELECT Mission_Outcome, COUNT(*) AS Count FROM SPACEXTABLE GROUP BY Mission_Outcome")
```

## Task 8 - Booster versions that carried the maximum payload


```python
q("SELECT Booster_Version FROM SPACEXTABLE WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE)")
```

## Task 9 - 2015 failed drone-ship landing records


```python
q("SELECT substr(Date,6,2) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE substr(Date,1,4)='2015' AND Landing_Outcome = 'Failure (drone ship)'")
```

## Task 10 - Rank landing outcomes in the specified date interval


```python
q("SELECT Landing_Outcome, COUNT(*) AS Count FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Count DESC")
```

## Key verified results
- Unique sites: CCAFS LC-40, CCAFS SLC-40, KSC LC-39A, VAFB SLC-4E
- NASA (CRS) payload total: 45,596 kg
- F9 v1.1 average payload: about 2,534.67 kg
- First successful ground-pad landing: 2015-12-22

# EDA with Data Visualization - Completed


```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv")
sns.set_theme(style="whitegrid")
```

## Flight Number vs Launch Site


```python
sns.catplot(data=df, x="FlightNumber", y="LaunchSite", hue="Class", aspect=2.5)
plt.title("Flight Number vs Launch Site")
plt.show()
```

## Payload Mass vs Launch Site


```python
sns.catplot(data=df, x="PayloadMass", y="LaunchSite", hue="Class", aspect=2.5)
plt.title("Payload Mass vs Launch Site")
plt.show()
```

## Success Rate vs Orbit Type


```python
orbit_success = df.groupby("Orbit", as_index=False)["Class"].mean()
sns.barplot(data=orbit_success, x="Orbit", y="Class")
plt.xticks(rotation=45)
plt.ylabel("Success Rate")
plt.title("Success Rate vs Orbit Type")
plt.show()
```

## Flight Number vs Orbit Type


```python
sns.catplot(data=df, x="FlightNumber", y="Orbit", hue="Class", aspect=2.5)
plt.title("Flight Number vs Orbit Type")
plt.show()
```

## Payload Mass vs Orbit Type


```python
sns.catplot(data=df, x="PayloadMass", y="Orbit", hue="Class", aspect=2.5)
plt.title("Payload Mass vs Orbit Type")
plt.show()
```

## Launch Success Yearly Trend


```python
df["Year"] = pd.to_datetime(df["Date"]).dt.year
yearly = df.groupby("Year", as_index=False)["Class"].mean()
sns.lineplot(data=yearly, x="Year", y="Class", marker="o")
plt.ylabel("Average Success Rate")
plt.title("Launch Success Yearly Trend")
plt.show()
```

## Findings
Landing success becomes more common at later flight numbers, varies by orbit and launch site, and improves strongly over time. Payload alone does not fully explain landing outcome.

# Predictive Analysis (Classification) - Completed


```python
import pandas as pd, numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
features = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_3.csv")
labels = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv")["Class"]
X = StandardScaler().fit_transform(features.astype(float))
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2, random_state=2, stratify=labels)
len(y_test)
```

## Logistic Regression


```python
lr = GridSearchCV(LogisticRegression(max_iter=1000), {"C":[0.01,0.1,1,10], "solver":["liblinear"]}, cv=10)
lr.fit(X_train,y_train)
lr.score(X_test,y_test), lr.best_score_, lr.best_params_
```

## Support Vector Machine


```python
svm = GridSearchCV(SVC(), {"kernel":["linear","rbf","sigmoid"], "C":[0.1,1,10], "gamma":["scale","auto"]}, cv=10)
svm.fit(X_train,y_train)
svm.score(X_test,y_test), svm.best_score_, svm.best_params_
```

## Decision Tree


```python
tree = GridSearchCV(DecisionTreeClassifier(random_state=2), {"criterion":["gini","entropy"], "max_depth":[2,4,6,8,10,None], "min_samples_split":[2,4,6], "max_features":[None,"sqrt","log2"]}, cv=10)
tree.fit(X_train,y_train)
tree.score(X_test,y_test), tree.best_score_, tree.best_params_
```

## K Nearest Neighbors


```python
knn = GridSearchCV(KNeighborsClassifier(), {"n_neighbors":list(range(1,11)), "weights":["uniform","distance"], "p":[1,2]}, cv=10)
knn.fit(X_train,y_train)
knn.score(X_test,y_test), knn.best_score_, knn.best_params_
```

## Compare models and confusion matrix


```python
models={"Logistic Regression":lr,"SVM":svm,"Decision Tree":tree,"KNN":knn}
for name,m in models.items():
    print(name, "test=", round(m.score(X_test,y_test),4), "cv=", round(m.best_score_,4), "params=", m.best_params_)
best=max(models, key=lambda k: models[k].best_score_)
pred=models[best].predict(X_test)
print("Best by CV:", best)
print(confusion_matrix(y_test,pred))
```
